# GMD 2017 Exploration

This notebook loads and reviews the `GMD_all_2017.dta` file.

The source file is large (about 3.5 GB), so the notebook uses 100,000-row chunks. `data` contains only a preview chunk; full-file summaries stream through the file without loading it all into memory. Run the cells from top to bottom.

```python
import pandas as pd

CHUNK_SIZE = 100_000
DATA_FILE = 'GMD_all_2017.dta'


def read_gmd_chunks():
    return pd.read_stata(DATA_FILE, iterator=True, chunksize=CHUNK_SIZE)
```

```python
data = next(read_gmd_chunks())
print(f'Preview chunk: {len(data):,} rows')
data.head()
```

```python
total_rows = 0
missing_values = None

for chunk in read_gmd_chunks():
    total_rows += len(chunk)
    chunk_missing = chunk.isna().sum()
    missing_values = chunk_missing if missing_values is None else missing_values.add(chunk_missing)

print(f'Rows: {total_rows:,}')
print(f'Columns: {data.shape[1]:,}')
```
