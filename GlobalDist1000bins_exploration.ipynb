# 1000 Binned Global Distribution

This notebook explores `GlobalDist1000bins_1990_2026_20260922_2021_01_02_PROD.csv` (about 770 MB). The file is read in 500,000-row chunks, so the full dataset is never loaded into memory at once.

```python
import pandas as pd

DATA_FILE = "GlobalDist1000bins_1990_2026_20260922_2021_01_02_PROD.csv"
CHUNK_SIZE = 500_000


def read_chunks():
    return pd.read_csv(DATA_FILE, chunksize=CHUNK_SIZE)
```

```python
preview = next(read_chunks())
print(f"Preview chunk: {len(preview):,} rows")
preview.head()
```

```python
total_rows = sum(len(chunk) for chunk in read_chunks())
print(f"Rows: {total_rows:,}")
print(f"Columns: {preview.shape[1]:,}")
preview.dtypes
```
